```python
import pandas as pd

def remove_date_column(input_file, output_file):
    # Cargar el CSV en un DataFrame
    df = pd.read_csv(input_file)
    
    # Verificar si la columna 'Date' existe y eliminarla
    if 'Date' in df.columns:
        df.drop(columns=['Date'], inplace=True)
    
    # Guardar el resultado en un nuevo archivo CSV
    df.to_csv(output_file, index=False)
    print(f"Archivo guardado sin la columna 'Date': {output_file}")

# Ejemplo de uso
input_csv = 'PremierLeague2324.csv'  # Nombre del archivo de entrada
output_csv = 'PremierLeague2324_Date.csv'  # Nombre del archivo de salida
remove_date_column(input_csv, output_csv)

```
